## TRAIN_TEST_SPLIT PRACTICE ON DIABETES DATASET ON 01 DEC 25

```python
import pandas as pd
```

```python
df = pd.read_csv('/Users/abhishek/Downloads/8. 100 Days ML from YT/Diabetes.csv')
df.shape
```

```python
df.columns
```

```python
# Dividing Columns for TRAINING and TESTING

X = df.drop(['diabetes'], axis = 1) 
# axis = 0 means rows, 1 means columns
```

```python
y = df['diabetes']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    train_size=0.7,
    stratify=y,
    random_state=2529
)
```

```python
X_train, X_test, y_train, y_test.shape
```

## Training of Model

```python
from sklearn.linear_model import LogisticRegression
```

```python
# Creating Model

model = LogisticRegression()
```

```python
# Training (Fit) the Model

model.fit(X_train, y_train)
```

```python
# Predicting test data

y_pred = model.predict(X_test)
```

```python
y_pred
```

## Testing performance and accuracy of Model

```python
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```
